```python
#| label: fig-distance-vs-displacement
#| fig-cap: 'Distance vs displacement: path length vs straight-line change in position'
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(8, 5))

# Draw curved path (distance)
t_path = np.linspace(0, 4*np.pi, 200)
x_path = t_path * 0.8
y_path = 2 * np.sin(t_path) + 3
ax.plot(x_path, y_path, 'b-', linewidth=2.5, label='Path travelled (distance)')

# Start and end points
ax.plot([x_path[0], x_path[-1]], [y_path[0], y_path[-1]], 'r--', linewidth=2, 
        label='Displacement (straight line)')
ax.plot(x_path[0], y_path[0], 'go', markersize=8, label='Start')
ax.plot(x_path[-1], y_path[-1], 'ro', markersize=8, label='Finish')

# Add distance and displacement labels
ax.text(5, 5, f'Distance ≈ {np.trapz(np.sqrt(np.diff(x_path)**2 + np.diff(y_path)**2), x_path[1:]):.1f} m', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.5))
ax.text(6, 2.5, f'Displacement = {np.sqrt((x_path[-1]-x_path[0])**2 + (y_path[-1]-y_path[0])**2):.1f} m', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightcoral', alpha=0.5))

# Axes and labels
ax.set_xlabel('Position x / m', fontsize=11)
ax.set_ylabel('Position y / m', fontsize=11)
ax.set_title('Distance vs Displacement', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10, loc='upper left')
ax.set_aspect('equal')
ax.set_xlim(-1, 12)
ax.set_ylim(-1, 6)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-acceleration-concept
#| fig-cap: 'Acceleration: change in velocity over time shown on velocity–time graph'
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

# Generate velocity-time data for constant acceleration
t = np.linspace(0, 8, 50)
u = 2  # initial velocity (m/s)
a = 1.5  # acceleration (m/s²)
v = u + a * t

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, v, 'b-', linewidth=2.5, label=f'v = u + at\n(u={u}, a={a})')

# Highlight initial and final points
ax.plot([0, t[-1]], [u, v[-1]], 'ro', markersize=8, zorder=5)
ax.plot([0, t[-1]], [u, v[-1]], 'r--', linewidth=1, alpha=0.7)

# Draw acceleration triangle
ax.plot([0, t[-1], t[-1]], [u, u, v[-1]], 'gray', linestyle=':', linewidth=1)
ax.text(t[-1]/2, u - 1, f'$\\Delta t = {t[-1]:.1f}$ s', ha='center', fontsize=10)
ax.text(t[-1] + 0.5, (u + v[-1])/2, f'$\\Delta v = {v[-1]-u:.1f}$ m/s', va='center', fontsize=10)

# Calculate and display acceleration
calc_a = (v[-1] - u) / t[-1]
ax.text(3, 10, f'Acceleration = gradient = $\\frac{{\\Delta v}}{{\\Delta t}} = \\frac{{{v[-1]-u:.1f}}}{{{t[-1]:.1f}}} = {calc_a:.1f}$ m/s²', 
        fontsize=10, fontweight='bold', bbox=dict(boxstyle='round', facecolor='lightyellow', edgecolor='orange'))

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Velocity / m s$^{-1}$', fontsize=11)
ax.set_title('Acceleration from Velocity–Time Graph', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.set_xlim(0, 9)
ax.set_ylim(0, 15)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-distance-time
#| fig-cap: Distance–time graphs showing different types of motion
#| fig-align: center

# Generate data for different motion types
t = np.linspace(0, 10, 100)

# Rest: constant distance
s_rest = np.ones_like(t) * 20

# Constant speed: s = 5t
s_constant = 5 * t

# Accelerating: s = 0.5*a*t^2, a=1 m/s^2
s_accel = 0.5 * 1 * t**2

# Decelerating: starts fast, slows down
s_decel = 15 * t - 0.6 * t**2
s_decel = np.clip(s_decel, 0, None)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, s_rest, 'b-', linewidth=2, label='At rest')
ax.plot(t, s_constant, 'g-', linewidth=2, label='Constant speed')
ax.plot(t, s_accel, 'r-', linewidth=2, label='Accelerating')
ax.plot(t, s_decel, 'm--', linewidth=2, label='Decelerating')

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Distance / m', fontsize=11)
ax.set_title('Distance–Time Graphs: Motion Types', fontsize=12, fontweight='bold')
ax.legend(loc='upper left', fontsize=9)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 10)
ax.set_ylim(0, 60)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-gradient-calculation
#| fig-cap: Calculating speed from gradient of a distance–time graph
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

# Generate linear distance-time data (constant speed)
t = np.linspace(0, 8, 50)
s = 4 * t + 2  # speed = 4 m/s, initial distance = 2 m

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, s, 'b-', linewidth=2.5, label='Distance–time graph')

# Highlight two points for gradient calculation
t1, t2 = 2, 6
s1, s2 = 4*t1 + 2, 4*t2 + 2
ax.plot([t1, t2], [s1, s2], 'ro', markersize=8, zorder=5)
ax.plot([t1, t2], [s1, s2], 'r--', linewidth=1.5, alpha=0.7)

# Draw gradient triangle
ax.plot([t1, t2, t2], [s1, s1, s2], 'gray', linestyle=':', linewidth=1)
ax.text((t1+t2)/2, s1 - 3, f'$\\Delta t = {t2-t1}$ s', ha='center', fontsize=10)
ax.text(t2 + 0.5, (s1+s2)/2, f'$\\Delta s = {s2-s1}$ m', va='center', fontsize=10)

# Calculate and display gradient
gradient = (s2 - s1) / (t2 - t1)
ax.text(4, 25, f'Speed = gradient = $\\frac{{\\Delta s}}{{\\Delta t}} = \\frac{{{s2-s1}}}{{{t2-t1}}} = {gradient}$ m/s', 
        fontsize=11, fontweight='bold', bbox=dict(boxstyle='round', facecolor='lightyellow', edgecolor='orange'))

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Distance / m', fontsize=11)
ax.set_title('Finding Speed from Gradient', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.set_xlim(0, 8.5)
ax.set_ylim(0, 35)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-speed-time
#| fig-cap: Speed–time graph with area calculation for distance travelled
#| fig-align: center

# Generate speed-time  accelerate, constant, decelerate
t1 = np.linspace(0, 6, 30)
t2 = np.linspace(6, 15, 45)
t3 = np.linspace(15, 24, 45)

v1 = 6 * t1
v2 = np.ones_like(t2) * 36
v3 = 36 - 4 * (t3 - 15)

t_combined = np.concatenate([t1, t2, t3])
v_combined = np.concatenate([v1, v2, v3])

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t_combined, v_combined, 'b-', linewidth=2.5)

ax.fill_between(t1, v1, alpha=0.3, color='blue', label='Area = distance (triangle)')
ax.fill_between(t2, v2, alpha=0.3, color='green', label='Area = distance (rectangle)')
ax.fill_between(t3, v3, alpha=0.3, color='red', label='Area = distance (triangle)')

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Speed / m s$^{-1}$', fontsize=11)
ax.set_title('Speed–Time Graph: Area = Distance Travelled', fontsize=12, fontweight='bold')
ax.legend(loc='upper right', fontsize=9)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 24)
ax.set_ylim(0, 40)

ax.text(12, 38, 'Total area = 594 m', fontsize=10, 
        bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.4))

plt.tight_layout()
plt.show()
```

```python
#| label: fig-free-fall
#| fig-cap: 'Speed–time graphs for free fall: with and without air resistance'
#| fig-align: center

t = np.linspace(0, 10, 100)
g = 10
v_vacuum = g * t

v_terminal = 50
k = 0.2
v_air = v_terminal * (1 - np.exp(-k * t))

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, v_vacuum, 'b-', linewidth=2.5, label='Vacuum (no air resistance)')
ax.plot(t, v_air, 'r--', linewidth=2.5, label='With air resistance')

ax.axhline(y=v_terminal, color='red', linestyle=':', linewidth=1, alpha=0.7)
ax.text(7, 52, 'Terminal velocity', fontsize=9, color='red', 
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.7))

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Speed / m s$^{-1}$', fontsize=11)
ax.set_title('Free Fall: Effect of Air Resistance', fontsize=12, fontweight='bold')
ax.legend(loc='upper left', fontsize=9)
ax.grid(True, alpha=0.3)
ax.set_xlim(0, 10)
ax.set_ylim(0, 110)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-free-fall-height
#| fig-cap: 'Height vs time for free fall: s = ½gt² relationship'
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

# Generate free fall height data
t = np.linspace(0, 4, 100)
g = 10
s = 0.5 * g * t**2  # s = ½gt²

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, s, 'b-', linewidth=2.5, label=f'$s = \\frac{{1}}{{2}}gt^2$ (g = {g} m/s²)')

# Highlight key points
times = [1, 2, 3, 4]
for tm in times:
    ht = 0.5 * g * tm**2
    ax.plot(tm, ht, 'ro', markersize=6)
    ax.text(tm + 0.2, ht, f'({tm}s, {ht:.0f}m)', fontsize=9)

ax.set_xlabel('Time / s', fontsize=11)
ax.set_ylabel('Height fallen / m', fontsize=11)
ax.set_title('Free Fall: Distance vs Time (s = ½gt²)', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.set_xlim(0, 4.5)
ax.set_ylim(0, 85)

# Add annotation about constant acceleration
ax.text(2.5, 70, 'Constant acceleration:\nvelocity increases linearly,\ndistance increases quadratically', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.7))

plt.tight_layout()
plt.show()
```

```python
#| label: fig-vector-addition
#| fig-cap: 'Vector addition: resultant of two perpendicular forces'
#| fig-align: center
#| warning: false

import matplotlib.patches as mpatches

# Define two perpendicular vectors
F1 = 4  # N, along x-axis (east)
F2 = 3  # N, along y-axis (north)

# Calculate resultant magnitude and direction
R = np.sqrt(F1**2 + F2**2)
theta = np.degrees(np.arctan2(F2, F1))

# Create figure and axis
fig, ax = plt.subplots(figsize=(6, 6))

# Draw coordinate axes
ax.axhline(y=0, color='black', linewidth=0.5)
ax.axvline(x=0, color='black', linewidth=0.5)

# Draw F1 (horizontal, blue)
ax.arrow(0, 0, F1, 0, head_width=0.3, head_length=0.4, 
         fc='blue', ec='blue', linewidth=2, label=f'$F_1$ = {F1} N (east)')

# Draw F2 (vertical, green) - starting from tip of F1
ax.arrow(F1, 0, 0, F2, head_width=0.3, head_length=0.4, 
         fc='green', ec='green', linewidth=2, label=f'$F_2$ = {F2} N (north)')

# Draw resultant R (red, diagonal)
ax.arrow(0, 0, F1, F2, head_width=0.3, head_length=0.4, 
         fc='red', ec='red', linewidth=2.5, label=f'Resultant $R$ = {R:.1f} N')

# Draw dashed lines to complete parallelogram (gray)
ax.plot([F1, F1], [0, F2], 'gray', linestyle='--', linewidth=1)
ax.plot([0, F1], [F2, F2], 'gray', linestyle='--', linewidth=1)

# CORRECTED: Use Arc (not Circle) for angle indicator - compatible with matplotlib >= 3.8
angle_arc = mpatches.Arc((0, 0), width=1.6, height=1.6, angle=0, 
                         theta1=0, theta2=theta, color='orange', linewidth=2)
ax.add_patch(angle_arc)

# Add angle label
ax.text(1.0, 0.3, f'${theta:.1f}^{{\\circ}}$', fontsize=11, color='orange')

# Labels and styling
ax.set_xlabel('East / N', fontsize=10)
ax.set_ylabel('North / N', fontsize=10)
ax.set_title('Vector Addition: Perpendicular Forces', fontsize=12, fontweight='bold')
ax.legend(loc='upper right', fontsize=9)
ax.grid(True, alpha=0.3)
ax.set_aspect('equal')
ax.set_xlim(-0.5, 5)
ax.set_ylim(-0.5, 4)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-reaction-time-uncertainty
#| fig-cap: Effect of human reaction time on timing measurements
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

# Reaction time distribution (typical 0.3-0.5 s)
reaction_times = np.random.normal(0.4, 0.08, 1000)
reaction_times = np.clip(reaction_times, 0.2, 0.7)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Histogram of reaction times
ax1.hist(reaction_times, bins=30, edgecolor='black', alpha=0.7, color='skyblue')
ax1.axvline(np.mean(reaction_times), color='red', linestyle='--', linewidth=2, 
            label=f'Mean = {np.mean(reaction_times):.2f} s')
ax1.set_xlabel('Reaction time / s', fontsize=11)
ax1.set_ylabel('Frequency', fontsize=11)
ax1.set_title('Human Reaction Time Distribution', fontsize=11, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=10)

# Effect on single vs multiple oscillation timing
n_oscillations = np.array([1, 5, 10, 20, 50])
true_period = 1.5  # seconds
reaction_error = 0.4  # typical reaction time error

# Relative error = reaction_error / (n * true_period)
relative_error = reaction_error / (n_oscillations * true_period) * 100

ax2.plot(n_oscillations, relative_error, 'bo-', linewidth=2, markersize=8)
ax2.fill_between(n_oscillations, 0, relative_error, alpha=0.3, color='blue')
ax2.set_xlabel('Number of oscillations timed', fontsize=11)
ax2.set_ylabel('Relative error / %', fontsize=11)
ax2.set_title('Why Time Multiple Oscillations?', fontsize=11, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.set_xlim(0, 52)
ax2.set_ylim(0, 30)

# Annotation
ax2.text(30, 25, 'Timing 20 oscillations\nreduces error to ~1.3%', 
         fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.7))

plt.tight_layout()
plt.show()
```

```python
#| label: fig-sign-convention
#| fig-cap: 'Sign convention for vertical motion: upward positive, downward negative'
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots(figsize=(6, 8))

# Draw vertical axis
ax.axvline(x=0, ymin=0.1, ymax=0.95, color='black', linewidth=2)
ax.arrow(0, 0.9, 0, 0.08, head_width=0.15, head_length=0.1, fc='black', ec='black')
ax.text(0.3, 0.95, '+ve direction (upward)', fontsize=10, fontweight='bold')

# Ground level
ax.axhline(y=0.1, xmin=0.3, xmax=0.7, color='brown', linewidth=3)
ax.text(0, 0.05, 'Ground', ha='center', fontsize=9)

# Object at different positions with velocity/acceleration vectors
positions = [0.7, 0.5, 0.3]  # y-positions
labels = ['Moving up', 'At top', 'Moving down']
velocities = [0.15, 0, -0.15]  # arrow lengths
accelerations = [-0.1, -0.1, -0.1]  # all downward (gravity)

for pos, label, vel, acc in zip(positions, labels, velocities, accelerations):
    # Object
    ax.plot(0, pos, 'ro', markersize=10)
    ax.text(0.3, pos, label, fontsize=9)
    
    # Velocity vector
    if vel != 0:
        color = 'blue' if vel > 0 else 'red'
        ax.arrow(0.15, pos, 0, vel*0.8, head_width=0.1, head_length=0.08, 
                 fc=color, ec=color, linewidth=1.5)
        ax.text(0.25, pos + vel*0.4, 'v', fontsize=10, color=color, fontweight='bold')
    
    # Acceleration vector (always downward)
    ax.arrow(-0.15, pos, 0, acc*0.8, head_width=0.1, head_length=0.08, 
             fc='green', ec='green', linewidth=1.5)
    ax.text(-0.4, pos + acc*0.4, 'a = -g', fontsize=9, color='green', fontweight='bold')

# Legend
ax.text(-0.5, 0.8, 'Key:\n🔴 Object\n→ v (velocity)\n↓ a (acceleration)', 
        fontsize=9, bbox=dict(boxstyle='round', facecolor='white', edgecolor='gray'))

ax.set_xlim(-0.6, 0.6)
ax.set_ylim(0, 1)
ax.set_xticks([])
ax.set_yticks([])
ax.set_title('Sign Convention: Vertical Motion', fontsize=12, fontweight='bold', pad=20)

plt.tight_layout()
plt.show()
```

```python
#| label: fig-light-gate-setup
#| fig-cap: Light gate experimental setup for measuring speed and acceleration
#| fig-align: center
#| warning: false

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

fig, ax = plt.subplots(figsize=(9, 5))

# Draw track/ramp
track = plt.Polygon([[0, 0.2], [8, 0.2], [8, 0.4], [0, 0.4]], 
                    closed=True, fill=True, color='lightgray', edgecolor='black', linewidth=1)
ax.add_patch(track)

# Draw light gates
gate_positions = [2, 5]
for i, pos in enumerate(gate_positions, 1):
    # Gate frame
    rect = mpatches.Rectangle((pos-0.3, 0), 0.6, 1.2, 
                              facecolor='lightblue', edgecolor='blue', linewidth=2)
    ax.add_patch(rect)
    # Light beam (dashed line)
    ax.plot([pos, pos], [0.3, 0.9], 'red', linestyle='--', linewidth=2, label='Light beam' if i==1 else "")
    # Label
    ax.text(pos, 1.3, f'Gate {i}', ha='center', fontsize=10, fontweight='bold')
    ax.text(pos, -0.2, f'$t_{i}$', ha='center', fontsize=10)

# Draw trolley with interrupt card
trolley_x = 1.5
trolley = mpatches.Rectangle((trolley_x-0.4, 0.25), 0.8, 0.3, 
                             facecolor='orange', edgecolor='black', linewidth=1.5)
ax.add_patch(trolley)
# Interrupt card
card = mpatches.Rectangle((trolley_x-0.1, 0.55), 0.2, 0.3, 
                          facecolor='yellow', edgecolor='black', linewidth=1)
ax.add_patch(card)
ax.text(trolley_x, 0.9, 'Interrupt card\n(length = d)', ha='center', fontsize=9)

# Arrows for motion
ax.annotate('', xy=(1, 0.7), xytext=(0.5, 0.7),
            arrowprops=dict(arrowstyle='->', linewidth=1.5, color='green'))
ax.text(0.75, 0.8, 'Motion', fontsize=9, color='green')

# Calculation annotation
ax.text(6.5, 0.8, 'Speed at gate:\n$v = \\frac{d}{\\Delta t}$', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', edgecolor='orange'))
ax.text(6.5, 0.4, 'Acceleration:\n$a = \\frac{v_2 - v_1}{t_2 - t_1}$', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightgreen', edgecolor='green'))

# Labels
ax.set_xlabel('Distance along track / m', fontsize=11)
ax.set_ylabel('Height (not to scale)', fontsize=11)
ax.set_title('Light Gate Setup: Measuring Speed & Acceleration', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.2, axis='x')
ax.set_xlim(0, 9)
ax.set_ylim(-0.5, 1.5)
ax.set_yticks([])

# Legend for light beam (only add once)
handles, labels = ax.get_legend_handles_labels()
if labels:
    ax.legend([handles[0]], [labels[0]], fontsize=9, loc='upper right')

plt.tight_layout()
plt.show()
```

```python
#| label: fig-incline-analysis
#| fig-cap: 'Acceleration vs sin(θ) for inclined plane: verifying a = g sin θ'
#| fig-align: center
#| warning: false

import numpy as np
import matplotlib.pyplot as plt

# Simulated experimental data
np.random.seed(42)
angles_deg = np.array([5, 10, 15, 20])
angles_rad = np.radians(angles_deg)
sin_theta = np.sin(angles_rad)
g_true = 9.81
a_theoretical = g_true * sin_theta
a_measured = a_theoretical + np.random.normal(0, 0.15, size=len(angles_deg))  # Add small error

fig, ax = plt.subplots(figsize=(8, 5))

# Plot experimental data
ax.plot(sin_theta, a_measured, 'bo', markersize=8, label='Experimental data', zorder=5)

# Linear fit through origin
coeffs = np.polyfit(sin_theta, a_measured, 1)
fit_line = np.poly1d(coeffs)
sin_smooth = np.linspace(0, 0.35, 100)
ax.plot(sin_smooth, fit_line(sin_smooth), 'r-', linewidth=2, 
        label=f'Best fit: gradient = {coeffs[0]:.2f} m/s²')

# Theoretical line (g = 9.81)
ax.plot(sin_smooth, g_true * sin_smooth, 'g--', linewidth=1.5, 
        label=f'Theoretical: $g = {g_true}$ m/s²')

# Labels and annotations
ax.set_xlabel('$\sin \\theta$', fontsize=12)
ax.set_ylabel('Acceleration $a$ / m s$^{-2}$', fontsize=11)
ax.set_title('Inclined Plane: $a$ vs $\sin\\theta$', fontsize=12, fontweight='bold')
ax.grid(True, alpha=0.3)
ax.legend(fontsize=10)
ax.set_xlim(0, 0.36)
ax.set_ylim(0, 4)

# Add gradient comparison box
ax.text(0.25, 3.5, f'Exp. $g$ = {coeffs[0]:.2f}\nTheory $g$ = {g_true:.2f}\nDiff = {abs(coeffs[0]-g_true)/g_true*100:.1f}%', 
        fontsize=10, bbox=dict(boxstyle='round', facecolor='lightyellow', edgecolor='orange'))

plt.tight_layout()
plt.show()
```
